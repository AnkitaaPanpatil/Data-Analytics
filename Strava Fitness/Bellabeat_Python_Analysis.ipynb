```python
df["ActivityDate"] = pd.to_datetime(df["ActivityDate"])
```

```python
df.info()
```

```python
df["Weekday"] = df["ActivityDate"].dt.day_name()
```

```python
df["Month"] = df["ActivityDate"].dt.month_name()
```

```python
df["Year"] = df["ActivityDate"].dt.year
```

```python
print("Total Unique Users:", df["Id"].nunique())
```

```python
print("Average Daily Steps:", round(df["TotalSteps"].mean(), 2))
```

```python
print("Average Calories Burned:", round(df["Calories"].mean(), 2))
```

```python
print("Maximum Steps:", df["TotalSteps"].max())
print("Minimum Steps:", df["TotalSteps"].min())
```

```python
print("Maximum Calories:", df["Calories"].max())
print("Minimum Calories:", df["Calories"].min())
```

```python
activity = df[[
    "VeryActiveMinutes",
    "FairlyActiveMinutes",
    "LightlyActiveMinutes",
    "SedentaryMinutes"
]].mean()

print(activity)
```

```python
top_users = (
    df.groupby("Id")["TotalSteps"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

print(top_users)
```

```python
top_calories = (
    df.groupby("Id")["Calories"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

print(top_calories)
```

```python
weekday_steps = (
    df.groupby("Weekday")["TotalSteps"]
      .mean()
)

print(weekday_steps)
```

```python
weekday_calories = (
    df.groupby("Weekday")["Calories"]
      .mean()
)

print(weekday_calories)
```

```python
correlation = df["TotalSteps"].corr(df["Calories"])

print("Correlation:", correlation)
```

```python
summary = df.groupby("Id").agg({
    "TotalSteps": ["sum", "mean", "max"],
    "Calories": ["sum", "mean"],
    "SedentaryMinutes": "mean"
})

summary
```

```python
summary.to_csv("User_Performance_Summary.csv")
```

```python
import matplotlib.pyplot as plt
```

```python
weekday_order = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

weekday_steps = df.groupby("Weekday")["TotalSteps"].mean().reindex(weekday_order)

plt.figure(figsize=(8,5))
plt.bar(weekday_steps.index, weekday_steps.values)
plt.title("Average Steps by Weekday")
plt.xlabel("Weekday")
plt.ylabel("Average Steps")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
weekday_calories = df.groupby("Weekday")["Calories"].mean().reindex(weekday_order)

plt.figure(figsize=(8,5))
plt.bar(weekday_calories.index, weekday_calories.values)
plt.title("Average Calories by Weekday")
plt.xlabel("Weekday")
plt.ylabel("Average Calories")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(df["TotalSteps"], df["Calories"])
plt.title("Steps vs Calories")
plt.xlabel("Total Steps")
plt.ylabel("Calories Burned")
plt.tight_layout()
plt.show()
```

```python
activity = df[[
    "VeryActiveMinutes",
    "FairlyActiveMinutes",
    "LightlyActiveMinutes",
    "SedentaryMinutes"
]].mean()

plt.figure(figsize=(8,5))
plt.bar(activity.index, activity.values)
plt.title("Average Activity Minutes")
plt.xlabel("Activity Type")
plt.ylabel("Average Minutes")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
top_users = (
    df.groupby("Id")["TotalSteps"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,6))
plt.barh(top_users.index.astype(str), top_users.values)
plt.title("Top 10 Active Users")
plt.xlabel("Total Steps")
plt.ylabel("User ID")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
top_calories = (
    df.groupby("Id")["Calories"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,6))
plt.barh(top_calories.index.astype(str), top_calories.values)
plt.title("Top 10 Users by Calories Burned")
plt.xlabel("Total Calories")
plt.ylabel("User ID")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
