```python
genres_df = df.copy()

genres_df['genres'] = (
    genres_df['genres']
    .str.replace(r'[\[\]\']', '', regex=True)
    .str.split(',')
)

genres_df = genres_df.explode('genres')

# Important fix
genres_df = genres_df.reset_index(drop=True)

genres_df['genres'] = genres_df['genres'].str.strip()
```

```python
heatmap_data = pd.crosstab(
    genres_df['release_year'],
    genres_df['genres']
)

heatmap_data.head()
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    heatmap_data.tail(20),
    cmap='YlGnBu'
)

plt.title("Genre Releases by Year")
plt.show()
```

```python
credits = pd.read_csv("credits.csv")

print(credits.columns.tolist())

credits.head()
```

```python
print(credits.columns.tolist())
```

```python
directors = credits[
    credits['role'] == 'DIRECTOR'
]

top_directors = directors['name'].value_counts().head(10)

print(top_directors)
```

```python
plt.figure(figsize=(10,5))

top_directors.plot(kind='bar')

plt.title("Top 10 Directors")
plt.xlabel("Director")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.show()
```

```python
actors = credits[
    credits['role'] == 'ACTOR'
]

top_actors = actors['name'].value_counts().head(10)

print(top_actors)
```

```python
plt.figure(figsize=(10,5))

top_actors.plot(kind='bar')

plt.title("Top 10 Actors")
plt.xlabel("Actor")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.show()
```
